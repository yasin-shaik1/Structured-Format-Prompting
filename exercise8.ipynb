```python
# ============================================================
# EXPERIMENT 8: STRUCTURED OUTPUT PROMPTING
# Task: Compare Unstructured vs Structured Responses
# Groq API + Google Colab
# ============================================================

!pip install -q groq

import os
from getpass import getpass
from groq import Groq


# ------------------------------------------------------------
# STEP 1: Load Groq API Key
# ------------------------------------------------------------

groq_api_key = getpass("Enter your Groq API key: ")

if not groq_api_key:
    raise ValueError("GROQ_API_KEY not provided.")

os.environ["GROQ_API_KEY"] = groq_api_key

print("✅ Groq API key loaded successfully.")


# ------------------------------------------------------------
# STEP 2: Create Groq Client
# ------------------------------------------------------------

client = Groq(
    api_key=os.getenv("GROQ_API_KEY")
)

print("✅ Groq client initialized.")


# ------------------------------------------------------------
# STEP 3: Define Function to Call LLM
# ------------------------------------------------------------

def ask_llm(prompt):

    try:

        response = client.chat.completions.create(
            model="llama-3.3-70b-versatile",
            temperature=0.2,
            max_tokens=400,
            messages=[
                {
                    "role": "user",
                    "content": prompt
                }
            ]
        )

        return response.choices[0].message.content.strip()

    except Exception as e:

        return f"❌ Error: {e}"


# ============================================================
# STEP 4: Define Unstructured Prompt
# ============================================================

unstructured_prompt = """
Describe Python, Java, and C++ in a short paragraph.

For each language, briefly mention:
- Its typing style
- A primary use case
- Its approximate difficulty level
"""


# ============================================================
# STEP 5: Define Structured Prompt
# ============================================================

structured_prompt = """
Create a Markdown table comparing Python, Java, and C++.

The table MUST contain exactly these columns:

| Language | Typing | Primary Use Case | Difficulty |

Use these rules:

1. Include exactly one row for Python.
2. Include exactly one row for Java.
3. Include exactly one row for C++.
4. Typing must be Static or Dynamic.
5. Difficulty must be Beginner, Intermediate, or Advanced.
6. Do not use JSON.
7. Output only the Markdown table.
"""


# ============================================================
# STEP 6: Generate Unstructured Response
# ============================================================

unstructured_response = ask_llm(
    unstructured_prompt
)


# ============================================================
# STEP 7: Generate Structured Response
# ============================================================

structured_response = ask_llm(
    structured_prompt
)


# ============================================================
# STEP 8: Display Unstructured Output
# ============================================================

print("\n" + "=" * 70)
print("UNSTRUCTURED OUTPUT")
print("=" * 70)

print("\nPrompt:")
print(unstructured_prompt)

print("\nResponse:")
print(unstructured_response)


# ============================================================
# STEP 9: Display Structured Output
# ============================================================

print("\n" + "=" * 70)
print("STRUCTURED OUTPUT - MARKDOWN TABLE")
print("=" * 70)

print("\nPrompt:")
print(structured_prompt)

print("\nResponse:")
print(structured_response)


# ============================================================
# STEP 10: Basic Structure Validation
# ============================================================

print("\n" + "=" * 70)
print("STRUCTURE VALIDATION")
print("=" * 70)

required_columns = [
    "Language",
    "Typing",
    "Primary Use Case",
    "Difficulty"
]

languages = [
    "Python",
    "Java",
    "C++"
]

# Check columns
missing_columns = []

for column in required_columns:

    if column.lower() not in structured_response.lower():
        missing_columns.append(column)


# Check languages
missing_languages = []

for language in languages:

    if language.lower() not in structured_response.lower():
        missing_languages.append(language)


# Display validation results

if not missing_columns:
    print("✅ All required columns are present.")
else:
    print("❌ Missing columns:", missing_columns)


if not missing_languages:
    print("✅ Python, Java, and C++ are present.")
else:
    print("❌ Missing languages:", missing_languages)


# Check JSON
if structured_response.strip().startswith("{"):

    print("❌ JSON-like output detected.")

else:

    print("✅ Output is not JSON.")


# ============================================================
# STEP 11: Final Analysis
# ============================================================

print("\n" + "=" * 70)
print("ANALYSIS")
print("=" * 70)

print("""
UNSTRUCTURED OUTPUT:
The information is presented as normal text. It is easy to read as
a paragraph, but comparing individual properties of Python, Java,
and C++ requires more effort.

STRUCTURED OUTPUT:
The Markdown table organizes the information into rows and columns.
This makes it easier to compare typing style, primary use case,
and difficulty level.

ADVANTAGES OF STRUCTURED OUTPUT:
1. Better readability.
2. Faster comparison between items.
3. Consistent formatting.
4. Easier extraction of individual fields.
5. Useful for reports and documentation.

CONCLUSION:
Structured prompting gives the LLM explicit instructions about the
desired output format. Compared with unstructured text, a Markdown
table makes comparisons clearer and more efficient.
""")


print("✅ Experiment 8 completed successfully.")
```
